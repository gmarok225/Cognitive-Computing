```python
L = [10, 0, 20, 40, 10, 70, 0, 20, 20, 60]
scores = tuple(L[:8])
print("Tuple scores:", scores)
```

```python
highest = max(scores)
highest_index = scores.index(highest)

lowest = min(scores)
lowest_count = scores.count(lowest)

print("Highest score:", highest, "| Found at index:", highest_index)
print("Lowest score:", lowest, "| Appears", lowest_count, "times")
```

```python
reversed_scores = list(scores[::-1])
print("Reversed scores (as list):", reversed_scores)
```

```python
user_score = int(input("Enter a score to search for: "))

if user_score in scores:
    print(f"{user_score} found at index {scores.index(user_score)}")
else:
    print(f"{user_score} is not present in the tuple")
```

```python
scores[0]=100

try:
    scores[0] = 100
except TypeError as e:
    print("Error caught:", e)
```

```python
first_score, second_score, *remaining_scores = scores

print("First score:", first_score)
print("Second score:", second_score)
print("Remaining scores:", remaining_scores)
```

```python
import random

roll_number = 1024170226
random.seed(roll_number)

numbers = [random.randint(100, 900) for _ in range(100)]
print("Random numbers:", numbers)
```

```python
odd_numbers = [n for n in numbers if n % 2 != 0]
print("Odd numbers count:", len(odd_numbers))
print("Odd numbers:", odd_numbers)
```

```python
even_numbers = [n for n in numbers if n % 2 == 0]
print("Even numbers count:", len(even_numbers))
print("Even numbers:", even_numbers)
```

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_numbers = [n for n in numbers if is_prime(n)]
print("Prime numbers count:", len(prime_numbers))
print("Prime numbers:", prime_numbers)
```

```python
from collections import Counter

counts = Counter(numbers)
most_common_num, most_common_count = counts.most_common(1)[0]
print("Most frequent number:", most_common_num, "| Occurs:", most_common_count, "times")
```

```python
roll_number = "1024170226"
first_8_digits = [int(d) for d in roll_number[:8]]
print("First 8 digits:", first_8_digits)

A = {digit * 7 for digit in first_8_digits}
B = {digit * 9 for digit in first_8_digits}

print("Set A:", A)
print("Set B:", B)
```

```python
union_AB = A.union(B)
print("Union (A U B):", union_AB)
```

```python
intersection_AB = A.intersection(B)
print("Intersection (A n B):", intersection_AB)
```

```python
diff_A_B = A.difference(B) 
diff_B_A = B.difference(A) 
print("A - B:", diff_A_B)
print("B - A:", diff_B_A)
```

```python
symdiff_AB = A.symmetric_difference(B)
print("Symmetric difference:", symdiff_AB)
```

```python
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))
```

```python
X = int(input("Enter a value to remove from set A: "))
A.discard(X)
print("Set A after discard:", A)

# discard() is safer than remove() because discard() silently does nothing if X is not found in the set,
# whereas remove() raises a KeyError if X isn't present — so discard() avoids crashing your program on a missing value.
```

```python
my_dict = {
    "name": "Gurnoor",
    "roll_no": "1024170226",
    "branch": "Computer Science",
    "age": 20,
    "city": "Patiala"
}
print(my_dict)
```

```python
my_dict["location"] = my_dict.pop("city")
print("After renaming city -> location:", my_dict)
```

```python
my_dict["cgpa"] = 8.7   # replace with your actual CGPA
print("After adding cgpa:", my_dict)
```

```python
my_dict["age"] += 1
print("After incrementing age:", my_dict)
```

```python
dict_copy1 = my_dict.copy()
dict_copy2 = my_dict.copy()

removed_value = dict_copy1.pop("branch")
print("dict_copy1 after pop('branch'):", dict_copy1)
print("Value returned by pop():", removed_value)

del dict_copy2["branch"]
print("dict_copy2 after del:", dict_copy2)

# pop() returns the removed value (useful if you need it), while del simply deletes the key-value pair and returns nothing.
```

```python
for key, value in my_dict.items():
    print(f"{key} → {value}")
```

```python
if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email not found in dictionary — no email on record.")
```

```python
friend_dict = {
    "name": "Aryan Sharma",
    "roll_no": "1024170299",
    "branch": "Electronics",
    "age": 21,
    "city": "Chandigarh"
}

merged_dict = {**my_dict, **friend_dict}
print("Merged dictionary:", merged_dict)

# When both dictionaries share a key, the values from the SECOND dictionary (friend_dict) "win" and overwrite the first's values,
# since later keyword arguments in the unpacking order take precedence.
```

```python
string_only_dict = {k: v for k, v in my_dict.items() if isinstance(v, str)}
print("Only string-valued key-value pairs:", string_only_dict)
```
