Q1. Write a unique paragraph (5-6 sentences) about your favorite topic (e.g., sports,
technology, food, books, etc.).
1. Convert text to lowercase and remove punctua5on using re.
2. Tokenize the text into words and sentences.
3. Split using split() and word_tokenize() and compare how Python split and NLTK’s
word_tokenize() differ.
4. Remove stopwords (using NLTK's stopwords list).
5. Display word frequency distribu5on (excluding stopwords).

```python
import re
import nltk
import matplotlib.pyplot as plt

from nltk.probability import FreqDist

text = """My favorite topic is artificial intelligence because it makes technology more intelligent and useful. AI helps people solve problems, analyze information, and automate repetitive tasks. I enjoy learning how machine learning models discover patterns from data. Natural language processing allows computers to understand and process human language. I believe AI will create exciting opportunities in education, healthcare, and research. Learning about AI motivates me to explore new ideas and build innovative projects."""

# 1. Convert to lowercase and remove punctuation
clean_text = re.sub(r'[^\w\s]', '', text.lower())
print("Cleaned text:\n", clean_text)

# 2. Tokenize words and sentences without NLTK corpora
words = re.findall(r'\b\w+\b', clean_text)
sentences = re.split(r'(?<=[.!?])\s+', text.strip())

print("\nWord tokens:", words)
print("\nSentence tokens:", sentences)

# 3. Compare split() and word_tokenize()
print("\nPython split():", text.split())
print("\nNLTK word_tokenize() requires tokenizer resources.")

# 4. Remove stopwords (using NLTK's stopwords list).
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in words
    if word not in stop_words
]

print("Words after removing stopwords:")
print(filtered_words)

# 5. Display word frequency distribu5on (excluding stopwords).
from nltk.probability import FreqDist
import matplotlib.pyplot as plt

freq_dist = FreqDist(filtered_words)

print("Word Frequency Distribution:")
for word, frequency in freq_dist.most_common():
    print(f"{word}: {frequency}")

plt.figure(figsize=(10, 5))
freq_dist.plot(15, title="Word Frequency Distribution (Without Stopwords)")
plt.show()
```

Q2. Using the same paragraph from Q1:
1. Extract all words with only alphabets using re.findall()
2. Remove stop words using NLTK’s stopword list
3. Perform stemming with PorterStemmer
4. Perform lemma5za5on with WordNetLemma5zer
5. Compare the stemmed and lemma5zed outputs and explain when you’d prefer one over
the other.

```python
import re

# Use the same paragraph from Q1
text = """My favorite topic is artificial intelligence because it makes technology more intelligent and useful. AI helps people solve problems, analyze information, and automate repetitive tasks. I enjoy learning how machine learning models discover patterns from data. Natural language processing allows computers to understand and process human language. I believe AI will create exciting opportunities in education, healthcare, and research. Learning about AI motivates me to explore new ideas and build innovative projects."""

# 1. Extract alphabetic words only and convert to lowercase
words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

print("Alphabetic words:")
print(words)

# 2. Remove stop words using NLTK’s stopword list
import nltk
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in words
    if word not in stop_words
]

print("Words after stopword removal:")
print(filtered_words)

# 3. Perform stemming with PorterStemmer
from nltk.stem import PorterStemmer

porter = PorterStemmer()

stemmed_words = [porter.stem(word) for word in filtered_words]

print("Stemmed words:")
print(stemmed_words)

# 4. Perform lemma5za5on with WordNetLemmatizer
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(word)
    for word in filtered_words
]

print("Lemmatized words:")
print(lemmatized_words)

# 5. Compare the stemmed and lemma5zed outputs and explain when you’d prefer one over the other.
print(f"{'Original':<18}{'Stemmed':<18}{'Lemmatized':<18}")
print("-" * 54)

for original, stemmed, lemmatized in zip(
    filtered_words, stemmed_words, lemmatized_words
):
    print(f"{original:<18}{stemmed:<18}{lemmatized:<18}")
```

Q3. Choose 3 short texts of your own (e.g., different news headlines, product reviews).
1. Use CountVectorizer to generate the Bag of Words representa5on.
2. Use TfidfVectorizer to compute TF-IDF scores.
3. Print and interpret the top 3 keywords from each text using TF-IDF.

```python
import sklearn

print("Scikit-learn version:", sklearn.__version__)

texts = [
    "Artificial intelligence improves technology and helps people solve problems.",
    "Healthy food provides energy and supports good health.",
    "Sports improve fitness and teach teamwork and discipline."
]
for i, text in enumerate(texts, start=1):
    print(f"Text {i}: {text}")

# 1. Use CountVectorizer to generate the Bag of Words representation.
from sklearn.feature_extraction.text import CountVectorizer

count_vectorizer = CountVectorizer()
bow_matrix = count_vectorizer.fit_transform(texts)

print("Vocabulary:")
print(count_vectorizer.get_feature_names_out())

print("\nBag of Words Matrix:")
print(bow_matrix.toarray())

# 2. Use TfidfVectorizer to compute TF-IDF scores.
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(texts)

print("Vocabulary:")
print(tfidf_vectorizer.get_feature_names_out())

print("\nTF-IDF Matrix:")
print(tfidf_matrix.toarray().round(3))


# 3. Print and interpret the top 3 keywords from each text using TF-IDF.
feature_names = tfidf_vectorizer.get_feature_names_out()

for i, row in enumerate(tfidf_matrix.toarray(), start=1):
    top_indices = row.argsort()[::-1][:3]

    print(f"\nTop 3 keywords in Text {i}:")
    for index in top_indices:
        print(f"{feature_names[index]}: {row[index]:.3f}")

```

Q4. Write 2 short texts (4–6 lines each) describing two different technologies (e.g., AI vs
Blockchain).
1. Preprocess and tokenize both texts.
2. Calculate:
a. Jaccard Similarity using sets
b. Cosine Similarity using TfidfVectorizer + cosine_similarity()
c. Analyze which similarity metric gives better insights in your case.

```python
text1 = """
Artificial intelligence enables machines to learn from data.
AI helps automate tasks and make intelligent decisions.
Machine learning improves predictions using experience.
AI is widely used in healthcare and education.
"""

text2 = """
Blockchain stores information in secure digital blocks.
It provides transparency and security for transactions.
Blockchain technology supports decentralized applications.
It is widely used in cryptocurrency and finance.
"""

print("Text 1: Artificial Intelligence")
print(text1)

print("\nText 2: Blockchain")
print(text2)


import re

def preprocess(text):
    text = text.lower()
    tokens = re.findall(r'\b[a-z]+\b', text)
    return tokens

tokens1 = preprocess(text1)
tokens2 = preprocess(text2)

print("Tokens in Text 1:", tokens1)
print("\nTokens in Text 2:", tokens2)


set1 = set(tokens1)
set2 = set(tokens2)

intersection = set1.intersection(set2)
union = set1.union(set2)

jaccard_similarity = len(intersection) / len(union)

print("Common words:", intersection)
print("Total unique words in union:", len(union))
print("Jaccard Similarity:", round(jaccard_similarity, 4))

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

texts = [text1, text2]

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(texts)

cosine_score = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[1:2])[0][0]

print("Cosine Similarity:", round(cosine_score, 4))

print("Jaccard Similarity:", round(jaccard_similarity, 4))
print("Cosine Similarity:", round(cosine_score, 4))

if cosine_score > jaccard_similarity:
    print("Cosine similarity indicates greater similarity because it considers TF-IDF word weights.")
elif jaccard_similarity > cosine_score:
    print("Jaccard similarity is higher for these texts.")
else:
    print("Both metrics give the same similarity score.")
```

Q5. Write a short review for a product or service.
1. Use TextBlob or VADER to find polarity & subjectivity for each review.
2. Classify reviews into Positive / Negative / Neutral.
3. Create a word cloud using the wordcloud library for all positive reviews.

```python
reviews = [
    "This smartphone has an excellent camera and amazing battery life. I absolutely love it.",
    "The laptop is fast, lightweight, and very easy to use. I am extremely happy with it.",
    "The headphones have wonderful sound quality and comfortable ear cushions.",
    "The phone battery is terrible and drains very quickly. I am disappointed.",
    "This laptop is slow, expensive, and frustrating to use.",
    "The headphones stopped working after two days. Very poor product.",
    "The product arrived yesterday.",
    "The package contains a phone charger.",
    "The device has a screen and two buttons."
]

for i, review in enumerate(reviews, start=1):
    print(f"Review {i}: {review}")

from textblob import TextBlob
import pandas as pd
import matplotlib.pyplot as plt
from wordcloud import WordCloud


results = []

for review in reviews:
    blob = TextBlob(review)

    results.append({
        "Review": review,
        "Polarity": blob.sentiment.polarity,
        "Subjectivity": blob.sentiment.subjectivity
    })

df = pd.DataFrame(results)

print(df.to_string(index=False))


def classify_sentiment(polarity):
    if polarity > 0:
        return "Positive"
    elif polarity < 0:
        return "Negative"
    else:
        return "Neutral"

df["Sentiment"] = df["Polarity"].apply(classify_sentiment)

print(df[["Review", "Polarity", "Subjectivity", "Sentiment"]].to_string(index=False))

print("\nSentiment counts:")
print(df["Sentiment"].value_counts())


positive_reviews = df[df["Sentiment"] == "Positive"]["Review"]

if len(positive_reviews) > 0:
    positive_text = " ".join(positive_reviews)

    wordcloud = WordCloud(
        width=1000,
        height=500,
        background_color="white",
        colormap="viridis"
    ).generate(positive_text)

    plt.figure(figsize=(12, 6))
    plt.imshow(wordcloud, interpolation="bilinear")
    plt.axis("off")
    plt.title("Word Cloud of Positive Product Reviews")
    plt.show()
else:
    print("No positive reviews available to generate a word cloud.")
```

Q6. Choose your own paragraph (~100 words) as training data.
1. Tokenize text using Tokenizer() from keras.preprocessing.text
2. Create input sequences and build a simple LSTM or Dense model
3. Train the model and generate 2–3 new lines of text star5ng from any seed word you
provide.

```python
text = """
Artificial intelligence is changing the world through innovation.
Machine learning helps computers learn patterns from data.
Natural language processing enables computers to understand human language.
Deep learning is used in image recognition and speech processing.
Artificial intelligence supports education, healthcare, and scientific research.
Data is important for building accurate machine learning models.
Neural networks learn complex patterns through repeated training.
Technology helps people solve problems and improve productivity.
Responsible artificial intelligence should be safe, fair, and reliable.
Learning artificial intelligence encourages creativity and innovation.
"""

print("Training text:\n", text)
print("Total words:", len(text.split()))

# Step 2: Tokenize the text using Keras Tokenizer()
from tensorflow.keras.preprocessing.text import Tokenizer

tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])

total_words = len(tokenizer.word_index) + 1

print("Word Index:")
print(tokenizer.word_index)
print("\nTotal unique words:", total_words)

# Step 3: Create input sequences
from tensorflow.keras.preprocessing.sequence import pad_sequences
import numpy as np

input_sequences = []

for sentence in text.lower().split("."):
    token_list = tokenizer.texts_to_sequences([sentence])[0]

    for i in range(1, len(token_list)):
        ngram_sequence = token_list[:i + 1]
        input_sequences.append(ngram_sequence)

max_sequence_len = max(len(seq) for seq in input_sequences)

input_sequences = pad_sequences(
    input_sequences,
    maxlen=max_sequence_len,
    padding="pre"
)

X = input_sequences[:, :-1]
y = input_sequences[:, -1]

print("Input shape:", X.shape)
print("Output shape:", y.shape)
print("Maximum sequence length:", max_sequence_len)

# Step 4: Build a simple LSTM model
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

model = Sequential([
    Embedding(input_dim=total_words, output_dim=32),
    LSTM(64),
    Dense(total_words, activation="softmax")
])

# Build the model explicitly
model.build(input_shape=(None, max_sequence_len - 1))

model.compile(
    loss="sparse_categorical_crossentropy",
    optimizer="adam",
    metrics=["accuracy"]
)

model.summary()

# Step 5: Train the model
history = model.fit(
    X,
    y,
    epochs=200,
    verbose=0
)

print("Model training completed!")
print("Final training accuracy:", round(history.history["accuracy"][-1], 4))

# Step 6: Generate text using a seed word
def generate_text(seed_text, next_words=8):
    for _ in range(next_words):
        token_list = tokenizer.texts_to_sequences([seed_text])[0]

        token_list = pad_sequences(
            [token_list],
            maxlen=max_sequence_len - 1,
            padding="pre"
        )

        predicted_probs = model.predict(token_list, verbose=0)[0]
        predicted_id = int(np.argmax(predicted_probs))

        output_word = tokenizer.index_word.get(predicted_id, "")

        if not output_word:
            break

        seed_text += " " + output_word

    return seed_text


# Generate 3 lines of text
seed_words = [
    "artificial intelligence",
    "machine learning",
    "technology"
]

for seed in seed_words:
    print(generate_text(seed, next_words=8))
```
